Rows and Columns Selection

```python

import pandas as pd

# ==================================================
# 1. CREATE A SUITABLE DATAFRAME
# ==================================================

data = {
    "Name": ["Rahul", "Priya", "Aman", "Neha", "Rohit"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 72, 95, 80],
    "City": ["Delhi", "Noida", "Delhi", "Ghaziabad", "Noida"]
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the complete DataFrame
print("ORIGINAL DATAFRAME:")
print(df)


# ==================================================
# 2. SELECT A SINGLE COLUMN
# ==================================================

# Select only the Name column
# Returns a Series
print("\n1. SINGLE COLUMN - NAME:")
print(df["Name"])


# ==================================================
# 3. SELECT MULTIPLE COLUMNS
# ==================================================

# Select Name and Marks columns
# Returns a DataFrame
print("\n2. MULTIPLE COLUMNS - NAME AND MARKS:")
print(df[["Name", "Marks"]])


# ==================================================
# 4. SELECT A SINGLE ROW USING iloc
# ==================================================

# Select the first row by integer position
# Positions start from 0
print("\n3. FIRST ROW USING iloc:")
print(df.iloc[0])


# ==================================================
# 5. SELECT MULTIPLE ROWS USING iloc
# ==================================================

# Select the first three rows
# The ending position 3 is excluded
print("\n4. FIRST THREE ROWS:")
print(df.iloc[0:3])


# ==================================================
# 6. SELECT A ROW USING loc
# ==================================================

# Select the row whose index label is 2
print("\n5. ROW WITH INDEX LABEL 2:")
print(df.loc[2])


# ==================================================
# 7. SELECT ROWS AND COLUMNS TOGETHER
# ==================================================

# Select the first row and Name and Marks columns
print("\n6. FIRST ROW - NAME AND MARKS:")
print(df.loc[0, ["Name", "Marks"]])

# Select the first three rows and first two columns
print("\n7. FIRST THREE ROWS AND FIRST TWO COLUMNS:")
print(df.iloc[0:3, 0:2])


# ==================================================
# 8. SELECT ROWS USING A CONDITION
# ==================================================

# Select students whose marks are greater than 85
print("\n8. STUDENTS WITH MARKS GREATER THAN 85:")
print(df[df["Marks"] > 85])


# ==================================================
# 9. SELECT ROWS USING MULTIPLE CONDITIONS
# ==================================================

# Select students who live in Delhi
# AND have marks greater than 80
print("\n9. DELHI STUDENTS WITH MARKS ABOVE 80:")
result = df[
    (df["City"] == "Delhi") &
    (df["Marks"] > 80)
]
print(result)


# ==================================================
# 10. SELECT MARKS GREATER THAN OR EQUAL TO 80
# ==================================================

# The >= operator also includes students scoring 80
print("\n10. STUDENTS WITH MARKS >= 80:")
print(df[df["Marks"] >= 80])
```

Filter DataFrame

```python

import pandas as pd

# 1. Create DataFrame
data = {
    "Name": ["Rahul", "Priya", "Aman", "Neha", "Rohit"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 72, 95, 80],
    "City": ["Delhi", "Noida", "Delhi", "Ghaziabad", "Noida"]
}

df = pd.DataFrame(data)

print("ORIGINAL DATAFRAME:")
print(df)


# 2. Filter marks greater than 80
print("\n1. MARKS > 80:")
print(df[df["Marks"] > 80])


# 3. Filter age less than 21
print("\n2. AGE < 21:")
print(df[df["Age"] < 21])


# 4. Filter students from Delhi
print("\n3. CITY IS DELHI:")
print(df[df["City"] == "Delhi"])


# 5. Filter marks greater than or equal to 80
print("\n4. MARKS >= 80:")
print(df[df["Marks"] >= 80])


# 6. AND: Both conditions must be True
print("\n5. DELHI AND MARKS > 80:")
print(df[(df["City"] == "Delhi") & (df["Marks"] > 80)])


# 7. OR: At least one condition must be True
print("\n6. DELHI OR NOIDA:")
print(df[(df["City"] == "Delhi") | (df["City"] == "Noida")])


# 8. Filter multiple specific cities using isin()
print("\n7. USING isin():")
print(df[df["City"].isin(["Delhi", "Noida"])])


# 9. Filter rows and select specific columns
print("\n8. SELECTED ROWS AND COLUMNS:")
print(df.loc[df["Marks"] >= 80, ["Name", "Marks"]])
```

Concaterate and Merge Dataframe(JOINS)


```python
df1=pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
df1

df2=pd.DataFrame({'ID':[1,2,3,4],'Score':[88,96,77,79]})
df2

print(pd.concat([df1,df2],axis=0)) #vertical/row level/top on top

print(pd.concat([df1,df2],axis=1)) # horizontal/column level/side on side
```

```python
#merge = join

pd.merge(df1,df2,how='outer',on='ID')
```

```python
pd.merge(df1,df2,how='outer',left_on='ID',right_on='ID')
```
